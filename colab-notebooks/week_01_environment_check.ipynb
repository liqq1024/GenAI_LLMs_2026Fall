#Step 4: Make your first LLM API call
Run the following cell to import the client and select a hosted instruction-following model:

```python
from huggingface_hub import InferenceClient
MODEL_ID = "meta-llama/Llama-3.1-8B-Instruct" #meta-llama/Llama-3.1-8B-Instruct
client = InferenceClient(
    model=MODEL_ID,
    token=HF_TOKEN,
    provider="auto"
    )
print(f"Client initialized for: {MODEL_ID}")
```

Next, send a prompt to the model:

```python
prompt = """ In 3 short bullet points, explain what a large language model is to a first-year college student. """
response = client.chat_completion(
    messages=[{"role": "user", "content": prompt} ],
    max_tokens=150,
    temperature=0.7
                                   )
print(response.choices[0].message.content)
```

The code sends your prompt to the hosted model and prints the generated response.

The two key parameters are:

*  *max_tokens*: the maximum length of the response.
*  *temperature*: the amount of randomness in the response. Lower values generally produce more focused and repeatable answers; higher values generally produce more varied answers.

Hugging Face’s InferenceClient is a Python interface for sending requests to supported hosted models. Reference: Hugging Face InferenceClient documentation

#Step 5: Compare parameter settings
Use the same prompt with two different settings. Run this cell:

```python
def generate_response(temperature, max_tokens):
  response = client.chat_completion(
      messages=[ {"role": "user", "content": prompt} ],
      temperature=temperature, max_tokens=max_tokens
      )
  return response.choices[0].message.content

focused_response = generate_response( temperature=0.2, max_tokens=60 )
creative_response = generate_response( temperature=1.0, max_tokens=200 )

print("=== Focused response: temperature=0.2, max_tokens=60 ===")
print(focused_response)

print("\n=== More variable response: temperature=1.0, max_tokens=200 ===")
print(creative_response)
```
